## 4. Use the agent from Python and look at the exposure heatmap

```python
from citypulse.agent import AgentSession
s = AgentSession("office", "bengaluru", use_live=True, persist_memory=False)
rep = s.run(lang="en")
print("Data source:", rep["data_source"], "-", rep["data_note"])
print(rep["summary"]["text"])
```

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots(1, 2, figsize=(13, 3.2), sharey=True)
for a, key, title in zip(ax, ("baseline", "after"), ("Baseline week", "After CityPulse plan")):
    im = a.imshow(rep[key]["grid"], aspect="auto", cmap="Oranges", vmin=0, vmax=max(max(r) for r in rep["baseline"]["grid"]))
    a.set_title(f"{title} - score {rep[key]['score']}"); a.set_xlabel("hour of day"); a.set_yticks(range(7))
ax[0].set_ylabel("day of the week (0 = first day)"); fig.colorbar(im, ax=ax); plt.show()
```
